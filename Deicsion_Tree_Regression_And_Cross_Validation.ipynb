```python
# Q: How do we import the necessary libraries for data manipulation and visualization?
# This cell imports pandas for data handling and matplotlib for plotting.
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Q: How do we fetch and reconstruct the Boston House Pricing Dataset since it was removed from scikit-learn?
# This cell downloads the raw Boston dataset from Carnegie Mellon University and parses it back into features and target arrays.
import pandas as pd
import numpy as np

data_url = "http://lib.stat.cmu.edu/datasets/boston"
raw_df = pd.read_csv(data_url, sep="\\s+", skiprows=22, header=None)

# Reconstruct the feature matrix (data) and the target vector (target)
data = np.hstack([raw_df.values[::2, :], raw_df.values[1::2, :2]])
target = raw_df.values[1::2, 2]

# Wrap them in a helper class to maintain compatibility with the original scikit-learn Bunch structure
class BostonDataset:
    def __init__(self, data, target):
        self.data = data
        self.target = target
        self.feature_names = ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT']

boston_df = BostonDataset(data, target)
```

```python
# Q: How do we verify that the dataset object has been successfully created?
# This displays the reconstructed boston_df object to confirm it is defined.
boston_df
```

```python
# Q: How do we organize our dataset into independent variables (features) and dependent variables (targets)?
# This cell creates a pandas DataFrame for the independent features (X) and extracts the target variable (y).
# independent features
X = pd.DataFrame(boston_df.data, columns=boston_df.feature_names)
# dependent features
y = boston_df.target
```

```python
# Q: How can we inspect the first few rows of our feature DataFrame to ensure the structure is correct?
# This displays the first 5 rows of the feature matrix X.
X.head()
```

```python
# Q: How do we split our dataset into training and testing sets to evaluate our model's performance?
# This cell splits the data: 67% for training and 33% for testing.
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
     X, y, test_size=0.33, random_state=42)
```

```python
# Q: How do we initialize a basic Decision Tree Regressor model?
# This imports and instantiates the Decision Tree Regressor.
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor()
```

```python
# Q: How do we train the Decision Tree Regressor on our training data?
# This cell fits (trains) the model using the training features and targets.
regressor.fit(X_train, y_train)
```

```python
# Q: How do we make predictions on unseen data using our trained model?
# This cell predicts house prices for the test feature set.
y_pred = regressor.predict(X_test)
```

```python
# Q: What do the predicted target values look like?
# This cell outputs the array of predicted prices for the test set.
y_pred
```

```python
# Q: How do we evaluate our baseline model's performance using the R-squared score?
# This calculates the R^2 metric to measure how well the baseline predictions match the actual targets.
from sklearn.metrics import r2_score
score = r2_score(y_pred, y_test)
```

```python
# Q: What is the R-squared score of our baseline Decision Tree Regressor?
# This prints the baseline model's R^2 score.
score
```

```python
# Q: How do we define the search space (grid of hyperparameters) for tuning our Decision Tree?
# This cell specifies hyperparameters to test during cross-validation, and instantiates a new regressor.
## Hyperparameter Tuning
parameter = {
 'criterion': ['squared_error', 'friedman_mse', 'absolute_error', 'poisson'],
  'splitter': ['best', 'random'],
  'max_depth': [1, 2, 3, 4, 5, 6, 7, 8, 10, 11, 12],
  'max_features': ['sqrt', 'log2'] # Note: 'auto' has been removed in newer versions
}
regressor = DecisionTreeRegressor()
```

```python
# Q: How do we set up Grid Search Cross-Validation to find the optimal combination of hyperparameters?
# This cell configures GridSearchCV to search the parameter space using 5-fold cross validation and MSE scoring.
from sklearn.model_selection import GridSearchCV
regressorcv = GridSearchCV(regressor, param_grid=parameter, cv=5, scoring='neg_mean_squared_error')
```

```python
# Q: How do we run the grid search process on our training data to find the best model configuration?
# This fits the Grid Search cross-validation on the training set.
regressorcv.fit(X_train, y_train)
```

```python
# Q: What are the best hyperparameter settings discovered by our Grid Search?
# This displays the optimal hyperparameter combination found.
regressorcv.best_params_
```

```python
# Q: How do we make predictions using the best estimator found during hyperparameter tuning?
# This uses the best cross-validated model to predict the test targets.
y_pred = regressorcv.predict(X_test)
```

```python
# Q: What is the final R-squared score of our tuned Decision Tree model?
# This computes and displays the R^2 score of the predictions made by the optimized model.
r2_score(y_pred, y_test)
```
